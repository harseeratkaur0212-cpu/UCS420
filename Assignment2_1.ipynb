```python
# Q1)
roll_no = "1024170092"

#(i)
L=[int(digit)*10 for digit in roll_no]
print("L is: ",L)

#(ii)
L.append(99)
print("After append:", L)
L.insert(3, 55)
print("After insert:", L)

#(iii)
L.remove(0)
print("After remove:", L)
L.pop(2)
print("After pop:", L)

#(iv)
L.sort()
print("Ascending:", L)
L.sort(reverse=True)
print("Descending:", L)

#(v)
print("First three:", L[:3])
print("Last three:", L[-3:])

#(vi)
avg = sum(L) / len(L)
above_avg = [x for x in L if x > avg]
print("Greater than average:", above_avg)
```

```python
#Q2)
L=[99, 90, 80, 50, 40, 30, 20, 20, 10, 0]
scores=tuple(L[:8])
print("Scores tuple:",scores)

#(i)
print("Max:",max(scores),"Index:", scores.index(max(scores)))
print("Min:",min(scores),"Count:", scores.count(min(scores)))

#(ii)
rev = list(scores[::-1])
print("Reversed:",rev)
# Tuples are immutable, so elements cannot be reordered in place.

#(iii)
val=int(input("Enter score: "))
print(f"Index:{scores.index(val)}"if val in scores else "Not present")

#(iv)
try:
    scores[0] = 100
except TypeError as e:
    print("Error:",e)
    # Tuples are immutable and unchangeable, unlike mutable lists.

#(v)
first,second,*remaining=scores
print(first,second,remaining)



```

```python
#Q3)
import random
roll_no = 1024170092
random.seed(roll_no)

#(i)
nums = [random.randint(100, 900) for _ in range(100)]
print("Generated numbers:", nums)

#(ii)
odds = [x for x in nums if x % 2 != 0]
print(f"Odd numbers count: {len(odds)}")

#(iii)
evens = [x for x in nums if x % 2 == 0]
print(f"Even numbers count: {len(evens)}")

#--------*-----------------*------------

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

#(iv)
primes=[x for x in nums if is_prime(x)]
print(f"Prime numbers count: {len(primes)}")
print("Primes list:",primes)

#(v)
most_frequent=max(nums,key=nums.count)
frequency=nums.count(most_frequent)
print(f"Most frequent number:{most_frequent} (appears {frequency} times)")
```

```python
#Q4)
roll_number_str = "1024170092"
digits = [int(d) for d in roll_number_str[:8]]
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}
print("Set A:", A)
print("Set B:", B)

#(vi)
print("Union (A | B):", A.union(B))

#(vii)
print("Intersection (A & B):", A.intersection(B))

#(viii)
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

#(ix)
print("Symmetric Difference:", A.symmetric_difference(B))

#(x)
print("Is A subset of B?:", A.issubset(B))
print("Is B superset of A?:", B.issuperset(A))

#(xi)
X = int(input("Enter value to remove from A: "))
A.discard(X)
print("Set A after discard:", A)

```
