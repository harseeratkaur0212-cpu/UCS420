```python
#Assignment 5.2 WAP to add all the prime numbers from 1 to n. n is given by user.
n = int(input("Enter the value of n: "))

sum = 0

for num in range(2, n + 1):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        sum = sum + num

print("Sum of prime numbers =", sum)
```

```python
# 6.2 prime number

def IsPrime(n):
	for i in range(2, n//2 + 1):
		if n%i==0:
			return 0
	return 1

print ("IsPrime(20)  --> ", IsPrime(20))
print ("IsPrime(23)  --> ", IsPrime(23))
print ("IsPrime(200) --> ", IsPrime(200))
print ("IsPrime(37)  --> ", IsPrime(37))
```

```python
#Assingment 6.2: WAP using function that add all prime numbers from 1 to n, n given by the user.

def AddPrime(n):
    s = 0
    for i in range(2, n + 1):
        prime = True
        for j in range(2, i):
            if i % j == 0:
                prime = False
                break
        if prime:
            s += i
    return s

n = int(input("Enter n: "))
print("Sum of prime numbers =", AddPrime(n))
```
