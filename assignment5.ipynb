```python
#Q1)
import numpy as np

# Create a 1D array with 5 elements
arr = np.array([10, 20, 30, 40, 50])

print("Original Array:", arr)

# a) Addition of 2 to all elements
print("After adding 2:", arr + 2)

# b) Multiply 3 with all elements
print("After multiplying by 3:", arr * 3)

# c) Divide every element by 2
print("After dividing by 2:", arr / 2)
```

```python
#Q2)
import numpy as np

# a) Reverse the array
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])


# b) Find most frequent value and its indices

# i. Array x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
max_count = np.max(counts)

print("\nFor array x:")
print("Most frequent value(s):", values[counts == max_count])

for value in values[counts == max_count]:
    print("Value", value, "indices:", np.where(x == value)[0])


# ii. Array y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
max_count = np.max(counts)

print("\nFor array y:")
print("Most frequent value(s):", values[counts == max_count])

for value in values[counts == max_count]:
    print("Value", value, "indices:", np.where(y == value)[0])
```

```python
#Q3)
import numpy as np

# Create 2-D array
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

# a) Access 1st row, 2nd column
print("1st row, 2nd column:", arr[0, 1])

# b) Access 3rd row, 1st column
print("3rd row, 1st column:", arr[2, 0])
```

```python
#Q4)
import numpy as np

# Create 1-D array with 25 evenly spaced numbers from 10 to 100
Harseerat = np.linspace(10, 100, 25)

print("Array:")
print(Harseerat)

# Array information
print("Dimensions:", Harseerat.ndim)
print("Shape:", Harseerat.shape)
print("Total elements:", Harseerat.size)
print("Data type:", Harseerat.dtype)
print("Total bytes:", Harseerat.nbytes)

# Transpose using reshape()
transpose = Harseerat.reshape(5, 5).T

print("\nTranspose using reshape():")
print(transpose)

# Transpose using T attribute
print("\nTranspose using T:")
print(Harseerat.T)
```

```python
#Q5)
import numpy as np

# Create 1-D NumPy array with 25 evenly spaced numbers from 10 to 100
Harseerat = np.linspace(10, 100, 25)

print("Array:")
print(Harseerat)

# Print array information
print("Dimensions:", Harseerat.ndim)
print("Shape:", Harseerat.shape)
print("Total elements:", Harseerat.size)
print("Data type:", Harseerat.dtype)
print("Total bytes:", Harseerat.nbytes)

# Transpose using reshape()
transpose = Harseerat.reshape(5, 5).T

print("\nTranspose using reshape():")
print(transpose)

# Transpose using T attribute
transpose_T = Harseerat.reshape(5, 5).T

print("\nTranspose using T attribute:")
print(transpose_T)
```
